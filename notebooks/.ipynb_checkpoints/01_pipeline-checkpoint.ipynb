# zt < -2: E[future residual >0]?

```python
residuals = results["residuals"]
zscores = results["zscores"]

horizons = [1, 5, 10, 20]

rows = []

for h in horizons:

    # Future residual return:
    # e_{t+1} + ... + e_{t+h}
    future_residual = sum(
        residuals.shift(-j)
        for j in range(1, h + 1)
    )

    long_mask = zscores < -2
    short_mask = zscores > 2

    long_future = future_residual.where(long_mask).stack()
    short_future = future_residual.where(short_mask).stack()

    rows.append({
        "horizon": h,

        "long_count": len(long_future),
        "long_mean_future_residual": long_future.mean(),

        "short_count": len(short_future),
        "short_mean_future_residual": short_future.mean()
    })

signal_test = pd.DataFrame(rows)

signal_test
```

```python
residuals = results["residuals"]
zscores = results["zscores"]

horizons = [1, 5, 10, 20]

rows = []

for h in horizons:

    future_residual = sum(
        residuals.shift(-j)
        for j in range(1, h + 1)
    )

    long_mask = zscores < -2
    short_mask = zscores > 2

    long_future = (
        future_residual
        .where(long_mask)
        .stack()
        .dropna()
    )

    short_future = (
        future_residual
        .where(short_mask)
        .stack()
        .dropna()
    )

    rows.append({
        "horizon": h,
        "long_count": long_mask.sum().sum(),
        "long_mean_future_residual": long_future.mean(),
        "short_count": short_mask.sum().sum(),
        "short_mean_future_residual": short_future.mean()
    })

signal_test = pd.DataFrame(rows)

signal_test
```

# fixing bags found short residual problem

```python
horizons = [1, 5, 10, 20]

rows = []

for h in horizons:

    # Future RAW stock returns
    future_raw = sum(
        stock_returns.shift(-j)
        for j in range(1, h + 1)
    )

    # Future PCA residual returns
    future_residual = sum(
        residuals.shift(-j)
        for j in range(1, h + 1)
    )

    long_mask = zscores < -2
    short_mask = zscores > 2

    long_raw = (
        future_raw
        .where(long_mask)
        .stack()
        .dropna()
    )

    short_raw = (
        future_raw
        .where(short_mask)
        .stack()
        .dropna()
    )

    long_residual = (
        future_residual
        .where(long_mask)
        .stack()
        .dropna()
    )

    short_residual = (
        future_residual
        .where(short_mask)
        .stack()
        .dropna()
    )

    rows.append({
        "horizon": h,

        "long_raw": long_raw.mean(),
        "long_residual": long_residual.mean(),

        "short_raw": short_raw.mean(),
        "short_residual": short_residual.mean()
    })

comparison = pd.DataFrame(rows)

comparison
```

# Pick z-entry=2.5

```python
results = run_backtest(
    stock_returns=stock_returns,
    pca_window=252,
    k=6,
    spread_window=20,
    z_window=252,
    entry_z=2.5,
    max_holding=20,
    position_size=0.05,
    max_positions=10,
    max_gross=0.50,
    cost_bps=10
)

performance = results["performance"]
```

```python
positions = results["positions"]
weights = results["weights"]

# Raw alpha weights BEFORE PCA neutralization
raw_weights = positions * 0.05

# Turnover caused only by entry / exit of alpha positions
raw_turnover = raw_weights.diff().abs().sum(axis=1)

# First day: compare against zero portfolio
raw_turnover.iloc[0] = raw_weights.iloc[0].abs().sum()

# Actual turnover AFTER PCA neutralization
actual_turnover = weights.diff().abs().sum(axis=1)
actual_turnover.iloc[0] = weights.iloc[0].abs().sum()

turnover_comparison = pd.DataFrame({
    "raw_signal_turnover": raw_turnover,
    "actual_pca_turnover": actual_turnover,
    "extra_pca_turnover": actual_turnover - raw_turnover
})

turnover_comparison.describe()
```

```python
print(
    "Average raw turnover:",
    raw_turnover.mean()
)

print(
    "Average actual turnover:",
    actual_turnover.mean()
)

print(
    "Extra turnover from PCA neutralization:",
    actual_turnover.mean() - raw_turnover.mean()
)
```
